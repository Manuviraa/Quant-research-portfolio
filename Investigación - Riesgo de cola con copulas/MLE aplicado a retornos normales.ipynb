## **Estimación de Máxima Verosimilitud aplicada a la distribución normal**

Aunque sabemos que los retornos del mercado no siguen una distribución normal, el aplicar MLE a esta presenta una buena oportunidad para empezar a practicar y entender la optimización para modelos de riesgo.

### **Calculando la función de  ln-verosimilitud**

Recordamos la función de probabilidad conjunta para nuestras variables aleatorias "X"

$$ f(x_1, x_2, \cdots, x_n | \theta) = P(X_1 = x_1, X_2 = x_2, \cdots, X_n = x_n | \theta)$$

Si estos son i.i.d, puede ser reescrita como:

$$P(X_1 = x_1, X_2 = x_2, \cdots, X_n = x_n | \theta) = P(X_1 = x_1 | \theta) \cdot P(X_2 = x_2 | \theta), \cdots, P(X_n = x_n | \theta) = \prod_{i=1}^{n} f(x_i | \theta) $$

### **Aplicandolo a la distribución normal**

Recordando la fórmula de la distribución normal

$$\frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)$$

Donde:

- $\sigma$: Es la desviación estandar
- $\mu$: Es la media

Podemos reescribir la función de densidad como:

$$f(x_i | \mu, \sigma) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x_i-\mu)^2}{2\sigma^2}\right)$$

Si un conjunto de datos es generado por una sola distribución normal, esta puede ser reescrita como:

$$\prod_{i=1}^{n} \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x_i-\mu)^2}{2\sigma^2}\right) $$

Se maximizara el logaritmo natural de esta función

$$(\hat\mu,\hat\sigma)=\arg\max_{\mu,\sigma}\ell(\mu,\sigma) =  \arg\max\ln \left[\prod_{i=1}^{n} \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x_i-\mu)^2}{2\sigma^2}\right)\right] =\sum_{i=1}^{n} \ln \left[\frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x_i-\mu)^2}{2\sigma^2}\right)\right]$$



Reescribimos esta función usando las propiedades de los logaritmos
$$\begin{aligned} 

& \sum_{i=1}^{n} \ln \left[\frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(x_i-\mu)^2}{2\sigma^2}\right)\right]

\\&= \sum_{i=1}^n \left[ \ln\left(\frac{1}{\sqrt{2\pi\sigma^2}}\right) - \frac{(x_i-\mu)^2}{2\sigma^2} \right]

\\&= \sum_{i=1}^{n} - \ln(\sqrt2\pi\sigma^2) -\frac{(x_i-\mu)^2}{2\sigma^2}

\\&= \sum_{i=1}^{n} - \ln(\sigma) - \ln(\sqrt{2\pi}) -\frac{(x_i-\mu)^2}{2\sigma^2}

\\&= \sum_{i=1}^{n} -\ln[(\sigma^2)^\frac{1}{2}] -\ln[(2\pi)^\frac{1}{2}] -\frac{(x_i-\mu)^2}{2\sigma^2}

\\&= \sum_{i=1}^{n} -\frac{1}{2} \ln(\sigma^2) - \frac{1}{2} \ln(2\pi) -\frac{(x_i-\mu)^2}{2\sigma^2}

\\&= \sum_{i=1}^{n} \left[-\frac12\ln(\sigma^2)\right] + \sum_{i=1}^n \left[-\frac12\ln(2\pi)\right] + \sum_{i=1}^n \left[-\frac{(x_i-\mu)^2}{2\sigma^2}\right]

\\&= -\frac{n}{2} \ln(\sigma^2) - \frac{n}{2} \ln(2\pi) -\frac{1}{2\sigma^2} \cdot \sum_{i=1}^{n} (x_i-\mu)^2
\end{aligned}$$
